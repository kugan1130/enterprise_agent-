# PostgreSQL Database Connection & Query Test

Tests PostgreSQL database connection (`SELECT 1`) and queries `SELECT COUNT(*) FROM sales;` using the central SQLAlchemy setup in `backend.app.core.database`.

```python
import sys
from pathlib import Path

# Resolve project root
cwd = Path.cwd().resolve()
if cwd.name == "tests":
    project_root = cwd.parent.parent
elif cwd.name == "backend":
    project_root = cwd.parent
else:
    project_root = cwd

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from sqlalchemy import text
from backend.app.core.database import SessionLocal

with SessionLocal() as session:
    conn_res = session.execute(text("SELECT 1;")).scalar()
    sales_count = session.execute(text("SELECT COUNT(*) FROM sales;")).scalar()
    print(f"1. Connectivity Test ('SELECT 1'): SUCCESS (Result: {conn_res})")
    print(f"2. Sales Table Record Count ('SELECT COUNT(*) FROM sales;'): {sales_count}")
```
